```python
#data

rankings=pd.read_csv("FIFA_Rankings.csv")
matches=pd.read_csv("world_cup_matches_scrape.csv")
full_gdp=pd.read_csv("WORLD_GDP_1960_PRESENT.csv", skiprows=3)
population=pd.read_csv("WORLD_POPULATION_1960_PRESENT.csv", skiprows=3)
```

```python
matches = pd.read_csv("world_cup_matches_scrape.csv")

#create match id
matches = matches.reset_index(drop=True)
matches["match_id"] = matches.index

matches
```

```python
#drop unwanted columns
gdp = full_gdp.drop(columns=["Unnamed: 70"])
pop = population.drop(columns=["Unnamed: 70"])

#convert long
gdp_long = gdp.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="GDP"
)

pop_long = pop.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="Population"
)

#clean year
gdp_long["Year"] = gdp_long["Year"].astype(int)
pop_long["Year"] = pop_long["Year"].astype(int)

gdp_long = gdp_long.drop(columns=["Indicator Name", "Indicator Code", "Country Name"])
pop_long = pop_long.drop(columns=["Indicator Name", "Indicator Code", "Country Name"])
```

```python
rankings["rank_date"] = pd.to_datetime(rankings["rank_date"])
rankings["Year"] = rankings["rank_date"].dt.year

#sort
rank = rankings.sort_values("rank_date")

#take first val
rank_clean = rank.groupby(
    ["country_abrv", "Year"],
    as_index=False
).first()
```

```python
#build target
def get_result(row):
    if row["home_score"] > row["away_score"]:
        return 1   # home win
    elif row["home_score"] < row["away_score"]:
        return -1  # away win
    else:
        return 0   # draw

matches["result"] = matches.apply(get_result, axis=1)

#only keep certain features
matches = matches[[
    "match_id","year","stage", "home_team", "away_team", "home_score", "away_score", "result"
]]
```

```python
country_map = gdp[["Country Name", "Country Code"]].drop_duplicates()
```

## Merging

```python
#merge home
matches = matches.merge(
    country_map,
    left_on="home_team",
    right_on="Country Name",
    how="left"
).rename(columns={"Country Code": "home_code"}).drop(columns=["Country Name"])

#merge away
matches = matches.merge(
    country_map,
    left_on="away_team",
    right_on="Country Name",
    how="left"
).rename(columns={"Country Code": "away_code"}).drop(columns=["Country Name"])

print(len(matches))
```

```python
#name mapping
country_fix = {
    "England": "GBR",
    "Scotland": "GBR",
    "Wales": "GBR",
    "Northern Ireland": "GBR",
    "United States": "USA",
    "South Korea": "KOR",
    "Iran": "IRN"
}

matches["home_code"] = matches["home_code"].fillna(
    matches["home_team"].map(country_fix)
)

matches["away_code"] = matches["away_code"].fillna(
    matches["away_team"].map(country_fix)
)
```

```python
#merge home gdp
matches = matches.merge(
    gdp_long,
    left_on=["home_code", "year"],
    right_on=["Country Code", "Year"],
    how="left"
).rename(columns={"GDP": "home_gdp"}).drop(columns=["Country Code", "Year"])

#home pop
matches = matches.merge(
    pop_long,
    left_on=["home_code", "year"],
    right_on=["Country Code", "Year"],
    how="left"
).rename(columns={"Population": "home_pop"}).drop(columns=["Country Code", "Year"])

#merge away gdp
matches = matches.merge(
    gdp_long,
    left_on=["away_code", "year"],
    right_on=["Country Code", "Year"],
    how="left"
).rename(columns={"GDP": "away_gdp"}).drop(columns=["Country Code", "Year"])

#merge away pop
matches = matches.merge(
    pop_long,
    left_on=["away_code", "year"],
    right_on=["Country Code", "Year"],
    how="left"
).rename(columns={"Population": "away_pop"}).drop(columns=["Country Code", "Year"])

print(len(matches))
```

```python
fifa_code_map = {
    "England": "ENG",
    "Scotland": "SCO",
    "Wales": "WAL",
    "Northern Ireland": "NIR",
    "United States": "USA",
    "South Korea": "KOR",
    "Iran": "IRN",
    "Algeria": "ALG",
    "Angola": "ANG",
    "Bulgaria": "BUL",
    "Chile": "CHI",
    "Costa Rica": "CRC",
    "Croatia": "CRO",
    "Czech Republic": "CZE",
    "Denmark": "DEN",
    "Egypt": "EGY",
    "FR Yugoslavia": "YUG",         
    "Germany": "GER",
    "Greece": "GRE",
    "Honduras": "HON",
    "Ivory Coast": "CIV",           
    "Netherlands": "NED",
    "North Korea": "PRK",
    "Paraguay": "PAR",
    "Portugal": "POR",
    "Republic of Ireland": "IRL",
    "Russia": "RUS",
    "Saudi Arabia": "KSA",
    "Serbia and Montenegro": "SCG", 
    "Slovakia": "SVK",
    "South Africa": "RSA",
    "Switzerland": "SUI",
    "Togo": "TOG",
    "Trinidad and Tobago": "TRI",
    "Turkey": "TUR",
    "Uruguay": "URU"
}

matches["home_fifa"] = matches["home_team"].map(fifa_code_map)
matches["away_fifa"] = matches["away_team"].map(fifa_code_map)

# fallback to existing codes where mapping not needed
matches["home_fifa"] = matches["home_fifa"].fillna(matches["home_code"])
matches["away_fifa"] = matches["away_fifa"].fillna(matches["away_code"])
```

```python


matches = matches.merge(
    rank_clean,
    left_on=["home_fifa", "year"],
    right_on=["country_abrv", "Year"],
    how="left"
).rename(columns={"rank": "home_rank"}).drop(columns=["country_abrv", "Year"])

matches = matches.merge(
    rank_clean,
    left_on=["away_fifa", "year"],
    right_on=["country_abrv", "Year"],
    how="left"
).rename(columns={"rank": "away_rank"}).drop(columns=["country_abrv", "Year"])


print(len(matches))
```

```python
#drop duplicates
matches = matches.loc[:, ~matches.columns.duplicated()]

#drop unwanted cols
matches = matches.drop(columns=[
    'country_full_x', 'total_points_x', 'previous_points_x',
    'rank_change_x', 'confederation_x', 'rank_date_x',
    'country_full_y', 'total_points_y', 'previous_points_y',
    'rank_change_y', 'confederation_y', 'rank_date_y'
], errors='ignore')
```

```python
final_df = matches[[
    "match_id",
    "year",
    "stage",
    "home_team",
    "away_team",
    "home_score",
    "away_score",
    "home_rank",
    "away_rank",
    "home_gdp",
    "away_gdp",
    "home_pop",
    "away_pop",
    "result"
]]
```

## Adding GDP and population data from Maddison Project 

```python
wc=final_df

maddison = pd.read_csv("maddison_data.csv")
#filter
maddison = maddison[maddison["year"] >= 1920]

#scale pop
maddison["pop"] = maddison["pop"] * 1000

#construct gdp
maddison["gdp"] = maddison["gdppc"] * maddison["pop"]

#remove unwanted
maddison = maddison[["country", "year", "gdp", "pop"]]
```

```python
mapping = {
    "England": "United Kingdom",
    "Scotland": "United Kingdom",
    "Wales": "United Kingdom",
    "Northern Ireland": "United Kingdom",
    "West Germany": "Germany",
    "East Germany": "Germany",
    "Soviet Union": "Russian Federation",
    "Yugoslavia": "Serbia",
    "FR Yugoslavia": "Serbia",
    "Serbia and Montenegro": "Serbia",
    "Czechoslovakia": "Czech Republic",
    "Republic of Ireland": "Ireland",
    "South Korea": "Republic of Korea",
    "North Korea": "D.P.R. of Korea",
    "Iran": "Iran (Islamic Republic of)",
    "Ivory Coast": "Côte d'Ivoire",
    "Zaire": "D.R. of the Congo",
    "Dutch East Indies": "Indonesia",
    "Bolivia": "Bolivia (Plurinational State of)",
    "Russia": "Russian Federation",
    "Yugoslavia": "Former Yugoslavia",
    "Soviet Union": "Former USSR",
}

wc["home_team_mapped"] = wc["home_team"].replace(mapping)
wc["away_team_mapped"] = wc["away_team"].replace(mapping)
```

```python
wc = wc.drop(columns=[
    "home_gdp", "away_gdp",
    "home_pop", "away_pop"
], errors="ignore")

wc["year"] = wc["year"].astype(int)
maddison["year"] = maddison["year"].astype(int)
```

```python
wc = wc.merge(
    maddison,
    left_on=["home_team_mapped", "year"],
    right_on=["country", "year"],
    how="left"
)

wc = wc.rename(columns={
    "gdp": "home_gdp",
    "pop": "home_pop"
}).drop(columns=["country"])

wc = wc.merge(
    maddison,
    left_on=["away_team_mapped", "year"],
    right_on=["country", "year"],
    how="left"
)

wc = wc.rename(columns={
    "gdp": "away_gdp",
    "pop": "away_pop"
}).drop(columns=["country"])

wc = wc.drop(columns=["home_team_mapped", "away_team_mapped"])
```

```python
print("Home GDP missing:", wc["home_gdp"].isna().mean())
print("Away GDP missing:", wc["away_gdp"].isna().mean())

print("Home Pop missing:", wc["home_pop"].isna().mean())
print("Away Pop missing:", wc["away_pop"].isna().mean())
wc
```
